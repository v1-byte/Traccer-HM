# Latih model digit HM Tracker di Google Colab (gratis)

Tanpa komputer: semuanya jalan di cloud Colab.

1. Di aplikasi: **Atur > Latih Kamera > Ekspor dataset (zip)** lalu kirim `hm_dataset.zip` ke akun Google Drive/HP yang sama dengan Colab.
2. Di Colab: menu **Runtime > Change runtime type > T4 GPU** (opsional, lebih cepat), lalu **Runtime > Run all**.
3. Saat diminta, pilih `hm_dataset.zip` dan `tools/train_digits.py` dari repo ini.
4. Unduh `hm_digits_model.zip` yang muncul di akhir, lalu di aplikasi pilih **Pasang model (zip)**.

Butuh 300+ sampel (skrip menolak di bawah 60). Bila ada galat versi TensorFlow, itu karena Colab memakai versi berbeda dari komputer pengembang; kirim pesan galatnya.

```
from google.colab import files
print('Pilih hm_dataset.zip')
files.upload()
print('Pilih train_digits.py')
files.upload()
```
